hero1 = Hero...? Coba lakukan print(hero1.hp).

Jawaban: Maka nilai hp pada objek hero1 berubah dari 100 menjadi 500. Karena hp adalah atribut yang bisa diubah setelah objek dibuat.

```python
class Hero:
  def __init__(self, name, hp, attack_power):
    self.name = name
    self.hp = hp
    self.attack_power = attack_power

  def info(self):
    print(f"Hero: {self.name} | HP: {self.hp} | Power:{self.attack_power}")

  # Method menyerang: Objek ini (self) menyerang objek lain (lawan)
  def serang(self, lawan):
    print(f"{self.name} menyerang {lawan.name}!")
    lawan.diserang(self.attack_power)

  # Method diserang: Menerima damage
  def diserang(self, damage):
    self.hp -= damage
    print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")

# -- Main Program --
hero1 = Hero("Layla", 100, 15)
hero2 = Hero("Zilong", 120, 20)

hero1.info()
hero2.info()

print("\n--- Pertarungan Dimulai ---")
hero1.serang(hero2)   # Layla menyerang Zilong
hero2.serang(hero1)   # Zilong membalas
```

Tugas Analisis 2:
Perhatikan parameter lawan pada method serang. Parameter tersebut
menerima sebuah objek utuh, bukan hanya string nama. Mengapa ini
penting?

Jawaban: Karena kalau cuma dikirim string nama, kita tidak bisa mengakses atau mengubah data lawan (seperti hp). Dengan mengirim objek utuh, method serang langsung punya akses ke semua atribut dan method milik lawan (lawan.hp, lawan.diserang()).

```python
class Mage(Hero):
  def __init__(self, name, hp, attack_power, mana):
    # Memanggil constructor milik Parent (Hero)
    super().__init__(name, hp, attack_power)
    self.mana = mana

  def info(self):
    print(f"{self.name} [Mage] | HP: {self.hp} | Mana: {self.mana}")

  # Mage punya skill khusus
  def skill_fireball(self, lawan):
    if self.mana >= 20:
      print(f"{self.name} menggunakan Fireball ke {lawan.name}!")
      self.mana -= 20
      lawan.diserang(self.attack_power * 2)  # Damage 2x lipat
    else:
      print(f"{self.name} gagal skill! Mana tidak cukup.")

# -- Main Program Baru --
print("\n--- Update Class Hero ---")
eudora = Mage("Eudora", 80, 30, 100)
balmond = Hero("Balmond", 200, 10)

eudora.info()
eudora.serang(balmond)         # Serangan biasa (warisan dari Hero)
eudora.skill_fireball(balmond) # Skill khusus Mage
```

Ketika kita menulis self.__hp di dalam class Hero, di belakang layar Python otomatis mengubah nama atribut itu menjadi _Hero__hp, mengikuti pola _NamaClass__namaAtribut. Proses inilah yang disebut name mangling. Jadi kalau kita tahu pola penamaannya, kita tetap bisa mengakses atribut itu dari luar class dengan menulis nama yang sudah "dimangling" tersebut.

Untuk uji validasi, kalau logika if dan elif di dalam set_hp dihapus sehingga isinya cuma self.__hp = nilai_baru, lalu kita panggil hero1.set_hp(-100), maka HP Hero akan langsung berubah jadi -100. Nilai negatif itu tersimpan begitu saja tanpa ada yang mencegah, padahal secara logika permainan HP negatif itu tidak masuk akal, seharusnya paling rendah ya 0 yang berarti karakter itu mati.

```python
from abc import ABC, abstractmethod

# 1. Interface / Abstract Class
# Ini adalah KONTRAK. Semua turunan wajib punya method di bawah ini.

class GameUnit(ABC):

  @abstractmethod
  def serang(self, target):
    pass

  @abstractmethod
  def info(self):
    pass

# 2. Implementasi pada Class Konkret
class Hero(GameUnit):
  def __init__(self, nama):
    self.nama = nama

# Kita WAJIB membuat method serang, kalau tidak akan Error
  def serang(self, target):
    print(f"Hero {self.nama} menebas {target}!")
  def info(self):
    print(f"Saya adalah Hero: {self.nama}")

class Monster(GameUnit):
  def __init__(self, jenis):
    self.jenis = jenis

# Implementasi serang versi Monster
  def serang(self, target):
    print(f"Monster {self.jenis} menggigit {target}!")
  def info(self):
    print(f"Saya adalah Monster: {self.jenis}")

# -- Uji Coba --
# unit = GameUnit() # ERROR! Abstract class tidak bisa jadi objek.
h = Hero("Alucard")
m = Monster("Serigala")
h.info()
m.info()
```

Tugas Analisis 5: Kalau method serang di Hero dihapus, program akan error dengan pesan TypeError: Can't instantiate abstract class Hero with abstract method serang.

Artinya, Python menolak membuat objek Hero karena Hero belum memenuhi "janji" dari GameUnit yang mewajibkan semua turunannya punya method serang. Errornya muncul langsung saat objek dibuat (h = Hero(...)), bukan saat method itu dipanggil belakangan.

Konsekuensinya kalau kita lupa membuat method yang dijanjikan interface: program langsung berhenti di awal (fail-fast), jadi kesalahan lebih cepat ketahuan daripada baru error di tengah jalan saat program sudah berjalan lama, yang biasanya lebih susah dilacak.

unit = GameUnit() juga akan error, karena GameUnit adalah abstract class yang cuma berisi method kosong (pass) sebagai janji, bukan implementasi nyata. Kalau dibiarkan bisa dibuat objek, method-nya tidak akan melakukan apa-apa yang berguna, dan itu berbahaya karena programmer lain bisa mengira unit itu berfungsi padahal tidak.

Gunanya GameUnit meski tidak bisa jadi objek: dia berfungsi sebagai standar/kontrak yang memaksa semua turunan (Hero, Monster) punya method serang dan info, walau implementasinya beda-beda. Ini bikin semua objek turunannya bisa diperlakukan secara seragam oleh kode lain — inti dari abstraksi.

```python
# Parent Class
class Hero:
  def __init__(self, nama):
    self.nama = nama

  def serang(self):
    print("Hero menyerang dengan tangan kosong.")
# Child Class 1
class Mage(Hero):
  def serang(self):
    print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")
# Child Class 2
class Archer(Hero):
  def serang(self):
    print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")
# Child Class 3
class Fighter(Hero):
  def serang(self):
    print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")
# -- Penerapan Polymorphism --
# Kita punya daftar hero campuran
pasukan = [
  Mage("Eudora"),
  Archer("Miya"),
  Fighter("Zilong"),
  Mage("Gord")
]
print("--- PERANG DIMULAI ---")

# Satu perintah loop, tapi respon berbeda-beda (Polymorphism)
for pahlawan in pasukan:
  pahlawan.serang()
```
