```python
from abc import ABC, abstractmethod

def format_rupiah(nominal):
    return f"Rp {nominal:,.0f}".replace(",", ".")

class KamarHotel(ABC):
    def __init__(self, nama_kamar, tipe, harga_dasar, stok=0):
        self.nama_kamar = nama_kamar
        self.tipe = tipe
        self.__harga_dasar = harga_dasar   # Private attribute
        self.__stok = stok                 # Private attribute

    # ---------- Getter ----------
    def get_stok(self):
        return self.__stok

    def get_harga_dasar(self):
        return self.__harga_dasar

    # ---------- Setter dengan validasi ----------
    def tambah_stok(self, jumlah):
        stok_baru = self.__stok + jumlah
        if stok_baru < 0:
            print(f"Gagal update stok {self.tipe}! Stok tidak boleh negatif ({jumlah}).")
            return False
        self.__stok = stok_baru
        print(f"Berhasil menambahkan stok {self.tipe}: {jumlah} unit.")
        return True

    # ---------- Abstract Methods ----------
    @abstractmethod
    def tampilkan_detail(self, nomor, jumlah_malam):
        pass

    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass

class KamarDeluxe(KamarHotel):
    PAJAK_PERSEN = 0.10
    def __init__(self, nama_kamar, harga_dasar, fasilitas, stok=0):
        super().__init__(nama_kamar, "Kamar Deluxe", harga_dasar, stok)
        self.fasilitas = fasilitas

    def hitung_harga_total(self, jumlah_malam):
        harga_dasar = self.get_harga_dasar()
        pajak = harga_dasar * self.PAJAK_PERSEN
        return (harga_dasar + pajak) * jumlah_malam

    def tampilkan_detail(self, nomor, jumlah_malam):
        harga_dasar = self.get_harga_dasar()
        pajak = harga_dasar * self.PAJAK_PERSEN
        subtotal = self.hitung_harga_total(jumlah_malam)

        print(f"{nomor}. [DELUXE] {self.nama_kamar} | Fasilitas: {self.fasilitas}")
        print(f"Harga Dasar/Malam: {format_rupiah(harga_dasar)} | "
              f"Pajak({int(self.PAJAK_PERSEN * 100)}%): {format_rupiah(pajak)}")
        print(f"Menginap: {jumlah_malam} malam | Subtotal: {format_rupiah(subtotal)}")
        return subtotal


class KamarStandard(KamarHotel):
    PAJAK_PERSEN = 0.05
    def __init__(self, nama_kamar, harga_dasar, kapasitas, stok=0):
        super().__init__(nama_kamar, "Kamar Standard", harga_dasar, stok)
        self.kapasitas = kapasitas

    def hitung_harga_total(self, jumlah_malam):
        harga_dasar = self.get_harga_dasar()
        pajak = harga_dasar * self.PAJAK_PERSEN
        return (harga_dasar + pajak) * jumlah_malam

    def tampilkan_detail(self, nomor, jumlah_malam):
        harga_dasar = self.get_harga_dasar()
        pajak = harga_dasar * self.PAJAK_PERSEN
        subtotal = self.hitung_harga_total(jumlah_malam)

        print(f"{nomor}. [STANDARD] {self.nama_kamar} | Kapasitas: {self.kapasitas}")
        print(f"Harga Dasar/Malam: {format_rupiah(harga_dasar)} | "
              f"Pajak({int(self.PAJAK_PERSEN * 100)}%): {format_rupiah(pajak)}")
        print(f"Menginap: {jumlah_malam} malam | Subtotal: {format_rupiah(subtotal)}")
        return subtotal

def proses_transaksi(daftar_pesanan):
    print("\n--- STRUK PEMESANAN ---")
    total_tagihan = 0
    for nomor, (kamar, jumlah_malam) in enumerate(daftar_pesanan, start=1):
        subtotal = kamar.tampilkan_detail(nomor, jumlah_malam)
        total_tagihan += subtotal
    print("----------------------------------------")
    print(f"TOTAL TAGIHAN: {format_rupiah(total_tagihan)}")
    print("----------------------------------------")
    return total_tagihan

if __name__ == "__main__":
    print("--- SETUP DATA KAMAR ---")

    deluxe = KamarDeluxe(
        nama_kamar="Kamar Deluxe Sea View",
        harga_dasar=1_500_000,
        fasilitas="Private Pool"
    )
    standard = KamarStandard(
        nama_kamar="Kamar Standard Superior",
        harga_dasar=500_000,
        kapasitas="2 Orang"
    )

    deluxe.tambah_stok(10)
    standard.tambah_stok(-5)     # ditolak karena hasil akhir negatif
    standard.tambah_stok(20)

    daftar_pesanan = [
        (deluxe, 2),      # Tamu memesan 2 malam Kamar Deluxe
        (standard, 1),    # Tamu memesan 1 malam Kamar Standard
    ]

    proses_transaksi(daftar_pesanan)
```
